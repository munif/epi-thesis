```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plotting(df, c1, c2):
    plt.figure(figsize=(10, 6))
    plt.scatter(df.index, df[c1], label=f"value_1 {c1}", marker='o',)
    plt.scatter(df.index, df[c2], label=f"value_1 {c2}", marker='x', alpha=0.5)
    
    plt.xlabel('Row Number')
    plt.ylabel('value_1')
    # plt.yscale('log', base=10)
    plt.title('y_train vs y_pred')
    plt.legend()
    plt.grid(True)
    
    # Save the plot to a file
    plt.savefig(f'img/fcnn/{c1}_vs_{c2}.png')
    
    plt.show()
```

```python
def plotting_selected(df, c1, c2):
    plt.figure(figsize=(10, 6))
    plt.scatter(df.index, df[c1], label=f"value_1 {c1}", marker='o',)
    plt.scatter(df.index, df[c2], label=f"value_1 {c2}", marker='x', alpha=0.5)
    
    plt.xlabel('Row Number')
    plt.ylabel('value_1')
    # plt.yscale('log', base=10)
    plt.title('y_train vs y_pred')
    plt.legend()
    plt.grid(True)
    
    # Save the plot to a file
    #plt.savefig(f'img/fcnn/{c1}_vs_{c2}.png')
    
    plt.show()
```

# Visualization for Train Dataset

```python
y_train = pd.read_csv('predictions/nn_train_predictions_20240805220133.csv')
y_train.head()
```

```python
y_train.describe()
```

```python
y_train.sort_values(by=y_train.columns[0], inplace=True)
y_train = y_train.reset_index(drop=True)
y_train.head()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(y_train.index, y_train['y_train'])
plt.show()
```

```python
plotting(y_train, 'y_train', 'y_pred')
```

```python
y_selected = y_train[(y_train['y_train'] <= 12.211925) & (y_train['y_pred'] <= 12.211925)]
y_selected.describe()
```

```python
y_selected.head()
```

```python
plotting_selected(y_selected, 'y_train', 'y_pred')
```
