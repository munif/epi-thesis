```python
# Read the Parquet file
df = pl.read_csv("gene_exp.csv")
```

```python
# Print the first few rows of the DataFrame
print(df.head())
```

```python
# Split the array column into separate columns
df_split = df.with_columns(
    [pl.col("h3k4me3").str.extract_all(r"\d")]
)
```

```python
df_split.head()
```

```python
def cast_list_to_float(lst):
    return [float(x) for x in lst]
```

```python
# Apply the function to the column
df_split = df_split.with_columns([
    pl.col("h3k4me3").map_elements(cast_list_to_float).alias("h3k4me3_float")
])
```
