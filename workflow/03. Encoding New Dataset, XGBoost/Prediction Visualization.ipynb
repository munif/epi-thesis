# Visualization of Test and Predicted Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datapath = "/group/pmc021/amunif/epi-thesis/scripts/shell/slurm/workflow/"
```

```python
dataset_path = "/group/pmc021/amunif/epi-thesis/dataset/"
```

```python
y_test = pd.read_csv(f"{datapath}y_test_10000.csv")
print(y_test.head())
print(y_test.shape)
```

```python
y_pred = pd.read_csv(f"{datapath}y_pred_10000.csv")
print(y_pred.head())
print(y_pred.shape)
```

```python
y_concat = pd.concat([y_test, y_pred], axis=1)
y_concat.head()
```

```python
y_concat.sort_values(by=['h_value_1', '0'], inplace=True)
y_concat.head()
```

```python
# Plotting
plt.figure(figsize=(10, 6))
plt.scatter(y_concat.index, y_concat['h_value_1'], label='y_test', marker='o')
plt.scatter(y_concat.index, y_concat['0'], label='y_pred', marker='x')

plt.xlabel('Row Number')
plt.ylabel('value_1')
plt.yscale('log', base=10)
plt.title('y_test vs y_pred')
plt.legend()
plt.grid(True)

# Save the plot to a file
plt.savefig('y_test_vs_y_pred.png')

plt.show()
```
