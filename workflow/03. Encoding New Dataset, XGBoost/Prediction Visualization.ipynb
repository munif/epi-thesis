# Visualization of Test and Predicted Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datapath = "/group/pmc021/amunif/epi-thesis/scripts/shell/slurm/workflow/"
```

```python
dataset_path = "/group/pmc021/amunif/epi-thesis/dataset/"
```

```python
y_test = pd.read_csv(f"{datapath}y_test.csv")
print(y_test.head())
print(y_test.shape)
```

```python
y_pred = pd.read_csv(f"{datapath}y_pred.csv")
print(y_pred.head())
print(y_pred.shape)
```

```python
# Plotting
plt.figure(figsize=(10, 6))
plt.scatter(y_test.index, y_test, label='y_test', marker='o')
plt.scatter(y_pred.index, y_pred, label='y_pred', marker='x')

plt.xlabel('Row Number')
plt.ylabel('value_1')
plt.yscale('log', base=10)
plt.title('y_test vs y_pred')
plt.legend()
plt.grid(True)
plt.show()
```
