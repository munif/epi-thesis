```python
import os
import glob
import pandas as pd
```

```python
OUTPUT_PATH = "/group/pmc021/amunif/epi-thesis/workflow/10_Permutation_Test_Ranking/output/E066v2/test/"
```

```python
# Get CSV files list from a folder
csv_files = glob.glob(os.path.join(OUTPUT_PATH, "*.csv"))
print(len(csv_files))
```

```python
# Read each CSV file into DataFrame
# This creates a list of dataframes
df_list = (pd.read_csv(file) for file in csv_files)
```

```python
# Concatenate all DataFrames
big_df = pd.concat(df_list, ignore_index=True)
```

```python
# Save the files into single file
big_df.to_csv("all_results_v2.csv", header=True, index=True)
```
