```python
import pandas as pd
import numpy as np
from itertools import chain, combinations, permutations
import os
import polars as pl
```

```python
WORKING_DIR = '/group/pmc021/amunif/epi-thesis/workflow/10_Permutation_Test_Ranking'
DATASET_DIR = '/group/pmc021/amunif/epi-thesis/workflow/10_Permutation_Test_Ranking/dataset'
```

```python
marker_list = ['H3K4me3', 'H3K9ac', 'H3K9me3', 'H3K27ac', 'H3K27me3']
```

```python
# Get the combination
def powerset(iterable):
    s = list(iterable)
    return chain.from_iterable(combinations(s, r) for r in range(1, len(s) + 1))
```

```python
# Get the marker powerset
markers_powerset = list(powerset(marker_list))
markers_powerset
```

```python
# Permutation of powerset
perm = []

for item in markers_powerset:
    perm.extend(list(permutations(item)))
```

```python
print(len(perm))
perm
```

```python
df = pd.DataFrame({'markers_perm': perm})
df
```

```python
# Save the permutation list
df.to_parquet(os.path.join(DATASET_DIR, 'permutation.parquet'))
```
