```python
import polars as pl
from pybiomart import Server
import os
import numpy as np
import pandas as pd
```

```python
WORKING_DIR = '/group/pmc021/amunif/epi-thesis/workflow/08_HepG2'
DATASET_DIR = '/group/pmc021/amunif/epi-thesis/dataset'
```

```python
# Load the matrix without label
gene_w_histone = pl.read_parquet(os.path.join(WORKING_DIR, 'dataset', 'gene_w_histone.parquet'))
```

```python
gene_w_histone
```

```python
# Load HepG2 expression dataset
hepg2_pl = pl.read_csv(os.path.join(DATASET_DIR, 'GSM3718064_HepG2_exp.txt'), separator="\t")
```

```python
hepg2_pl
```

```python
hepg2_pl = hepg2_pl.select(["gene_id", "value_1"])
```

```python
hepg2_pl
```

```python
# Join with histone dataset
gene_joined = gene_w_histone.join(hepg2_pl, on='gene_id')
```

```python
gene_joined
```

```python
# Convert into labels using quantile value
# Define the quartile boundaries
quartiles = [0.25, 0.5, 0.75]
```

```python
# Create the quartile labels
gene_w_label = gene_joined.with_columns([
    pl.col('value_1')
    .qcut(quartiles, labels=['0', '1', '2', '3'])
    .alias('label')
])
```

```python
gene_w_label
```

```python
gene_w_label = gene_w_label.with_columns(pl.col('label').cast(pl.Int64))
```

```python
gene_w_label
```

```python
# Check the label distribution
gene_w_label.group_by("label").len()
```

```python
# Save to parquet
gene_w_label.write_parquet(os.path.join(WORKING_DIR, 'dataset', f'gene_w_qlabel.parquet'))
```
