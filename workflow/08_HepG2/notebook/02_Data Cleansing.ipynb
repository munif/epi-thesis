```python
import polars as pl
from pybiomart import Server
import os
import numpy as np
import pandas as pd
```

```python
WORKING_DIR = '/group/pmc021/amunif/epi-thesis/workflow/08_HepG2'
DATASET_DIR = '/group/pmc021/amunif/epi-thesis/dataset'
```

# Loading Data

## Define schema and load data

```python
schema = pl.Schema({
    "Chromosome/scaffold name"  : pl.String(),
    "Gene start (bp)"           : pl.Int64(),
    "Gene end (bp)"             : pl.Int64(),
    "Strand"                    : pl.Int64(),
    "Gene name"                 : pl.String(),
    "Gene stable ID"            : pl.String(),
    "Gene type"                 : pl.String(),
    "tss"                       : pl.Int64(),
    "tss -2kb"                  : pl.Int64(),
    "tss +2kb"                  : pl.Int64(),
    "test_id"                   : pl.String(),
    "orig_gene_name"            : pl.String(),
    "locus"                     : pl.String(),
    "orig_start"                : pl.Int64(),
    "orig_end"                  : pl.Int64(),
})
```

```python
ensembl_pl = pl.read_csv(os.path.join(WORKING_DIR, "dataset", "ensembl.csv"), schema=schema)
```

```python
ensembl_pl
```

# Data Filtering

## Filter where gene name equals to original gene_name

```python
ensembl_match = ensembl_pl.filter(pl.col('Gene name') == pl.col('orig_gene_name'))
ensembl_match
```

```python
ensembl_match.write_csv(os.path.join(WORKING_DIR, "dataset", "ensembl_match.csv"))
```

## Gene with '-' which are not matched

```python
ensembl_without_name = ensembl_pl.unique(subset = ['locus', 'Gene name', 'test_id']).filter(pl.col('orig_gene_name') == '-')
ensembl_without_name
```

## All data without gene name checking

```python
ensembl_unique = ensembl_pl.unique(subset = ['locus', 'Gene name', 'test_id']).sort(['test_id']).select(pl.exclude(['orig_gene_name', 'orig_start', 'orig_end']))
```

```python
ensembl_unique
```

```python
ensembl_unique.write_csv(os.path.join(WORKING_DIR, "dataset", "ensembl_unique.csv"))
```

## Filter the data by gene type with 'protein_coding'

```python
ensembl_protein_coding = ensembl_pl.filter(pl.col('Gene type') == 'protein_coding')
ensembl_protein_coding
```
