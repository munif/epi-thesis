```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Evaluation Metric Visualization

```python
train_eval_metric = pd.read_csv('output-csv/mae/train_eval_metric.csv')
train_eval_metric.head()
```

```python
test_eval_metric = pd.read_csv('output-csv/mae/test_eval_metric.csv')
test_eval_metric.head()
```

```python
# MSE Graph
plt.figure(figsize=(10, 6))
plt.plot(train_eval_metric['n_estimators'], train_eval_metric['mse'], label='train_MSE')
plt.plot(train_eval_metric['n_estimators'], test_eval_metric['mse'], label='test_MSE')

plt.xlabel('n_estimators')
plt.ylabel('MSE')
plt.title('MSE Train vs Test')
plt.legend()
plt.grid(True)

# Save the plot to a file
plt.savefig('img/mse_train_vs_test.png')

plt.show()
```

```python
# RMSE Graph
plt.figure(figsize=(10, 6))
plt.plot(train_eval_metric['n_estimators'], train_eval_metric['rmse'], label='train_RMSE')
plt.plot(train_eval_metric['n_estimators'], test_eval_metric['rmse'], label='test_RMSE')

plt.xlabel('n_estimators')
plt.ylabel('RMSE')
plt.title('RMSE Train vs Test')
plt.legend()
plt.grid(True)

# Save the plot to a file
plt.savefig('img/rmse_train_vs_test.png')

plt.show()
```

```python
# MAE Graph
plt.figure(figsize=(10, 6))
plt.plot(train_eval_metric['n_estimators'], train_eval_metric['mae'], label='train_mae')
plt.plot(train_eval_metric['n_estimators'], test_eval_metric['mae'], label='test_mae')

plt.xlabel('n_estimators')
plt.ylabel('MAE')
plt.title('MAE Train vs Test')
plt.legend()
plt.grid(True)

# Save the plot to a file
plt.savefig('img/mae_train_vs_test.png')

plt.show()
```

```python
# R2 Graph
plt.figure(figsize=(10, 6))
plt.plot(train_eval_metric['n_estimators'], train_eval_metric['r2_score'], label='train_RMSE')
plt.plot(train_eval_metric['n_estimators'], test_eval_metric['r2_score'], label='test_RMSE')

plt.xlabel('n_estimators')
plt.ylabel('R2')
plt.title('R2 Train vs Test')
plt.legend()
plt.grid(True)

# Save the plot to a file
plt.savefig('img/r2_train_vs_test.png')

plt.show()
```

# Training and Testing Visualization

## Training

```python
y_train = pd.read_csv('output-csv/mae/y_train.csv').rename(columns={'h_value_1': 'y_train'})
y_train_pred_100 = pd.read_csv('output-csv/mae/xgboost_100_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_100'})
y_train_pred_200 = pd.read_csv('output-csv/mae/xgboost_200_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_200'})
y_train_pred_300 = pd.read_csv('output-csv/mae/xgboost_300_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_300'})
y_train_pred_500 = pd.read_csv('output-csv/mae/xgboost_500_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_500'})
y_train_pred_1000 = pd.read_csv('output-csv/mae/xgboost_1000_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_1000'})
y_train_pred_5000 = pd.read_csv('output-csv/mae/xgboost_5000_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_5000'})
y_train_pred_10000 = pd.read_csv('output-csv/mae/xgboost_10000_y_train_pred.csv').rename(columns={'h_value_1': 'y_pred_10000'})
```

```python
y = pd.concat([y_train, y_train_pred_100, y_train_pred_200, y_train_pred_300, 
              y_train_pred_500, y_train_pred_1000, y_train_pred_5000,
              y_train_pred_10000], axis=1)
y.head()
```

```python
y.sort_values(by=y.columns[0], inplace=True)
y.head()

# Check with the sorted value again
```

```python
y = y.reset_index(drop=True)
y.head()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(y.index, y['y_train'])
# plt.yscale('log', base=10)
plt.show()
```

```python
def plotting(df, c1, c2):
    plt.figure(figsize=(10, 6))
    plt.scatter(df.index, df[c1], label=f"value_1 {c1}", marker='o',)
    plt.scatter(df.index, df[c2], label=f"value_1 {c2}", marker='x', alpha=0.5)
    
    plt.xlabel('Row Number')
    plt.ylabel('value_1')
    # plt.yscale('log', base=10)
    plt.title('y_train vs y_pred')
    plt.legend()
    plt.grid(True)
    
    # Save the plot to a file
    plt.savefig(f'img/{c1}_vs_{c2}.png')
    
    plt.show()
```

```python
def plotting_selected(df, c1, c2):
    plt.figure(figsize=(10, 6))
    plt.scatter(df.index, df[c1], label=f"value_1 {c1}", marker='o',)
    plt.scatter(df.index, df[c2], label=f"value_1 {c2}", marker='x', alpha=0.5)
    
    plt.xlabel('Row Number')
    plt.ylabel('value_1')
    # plt.yscale('log', base=10)
    plt.title('y_train vs y_pred')
    plt.legend()
    plt.grid(True)
    
    # Save the plot to a file
    plt.savefig(f'img/selected/{c1}_vs_{c2}.png')
    
    plt.show()
```
