```python
import polars as pl
import os
```

```python
DATASET_PATH = "/group/pmc021/amunif/epi-thesis/workflow/13_Healthy liver with manual preprocessing/dataset/donor_3"
OUTPUT_PATH  = "/group/pmc021/amunif/epi-thesis/workflow/13_Healthy liver with manual preprocessing/output/donor_3/"
```

```python
### Merge the results into single CSV file

# Read all CSV into single dataframe
pl_df = pl.read_csv(os.path.join(OUTPUT_PATH, 'test', "*.csv"))
```

```python
pl_df
```

```python
# Add the histone number column
pl_df = pl_df.with_columns(
     (pl.col("histone_marker").str.count_matches("-") + 1).alias("number_of_histones")
)
```

```python
pl_df
```

```python
# Sort by number_of_histones (desc) and histone_marker (asc)
pl_df = pl_df.sort(
    by=["number_of_histones", "histone_marker"],
    descending=[True, False]
)
```

```python
pl_df
```

```python
# Add the rank column
pl_df = pl_df.with_columns(
    pl.col('test_accuracy').rank(method='ordinal', descending=True).alias('order_by_test_acc')
)

pl_df
```

```python
# Save the dataframe as CSV
pl_df.write_csv(os.path.join("Donor 3 ranking.csv"))
```
