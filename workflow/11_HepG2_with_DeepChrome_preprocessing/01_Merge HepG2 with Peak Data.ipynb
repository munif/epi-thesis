```python
def read_gappedPeak(filename):
    # Define the columns
    cols = [
        "chrom", "chromStart", "chromEnd", "name", "score", "strand",
        "thickStart", "thickEnd", "itemRgb", "blockCount", "blockSizes",
        "blockStarts", "signalValue", "pValue", "qValue"
    ]
    
    # Read the file
    gappedpeak_df = pl.read_csv(
        filename,
        separator="\t",
        has_header=False,
        new_columns=cols
    )
    
    # Return the dataframe
    return gappedpeak_df
```

```python
def read_histone(filename):
    # Define the columns
    cols = ["chrom", "chromStart", "chromEnd", "name"]
    
    # Read the histone file
    histone_df = pl.read_csv(
        filename,
        separator="\t",
        has_header=False,
        new_columns=cols
    )

    # Return the dataframe
    return histone_df
```

# Open the HepG2 peak dataset

```python
# Make an overlap join
H3K9ac_gp_simple_df = H3K9ac_gp_df.select(["chrom", "chromStart", "chromEnd", "signalValue", "pValue", "qValue"])
H3K9ac_simple_df = H3K9ac_df.select(["chrom", "chromStart", "chromEnd", "name"])
```

```python
H3K9ac_gp_simple_df
```

```python
H3K9ac_simple_df
```

```python
# Left join on chromosome
overlap_df = (
    H3K9ac_simple_df.join(
        H3K9ac_gp_simple_df,
        on="chrom",
        how="left",
        suffix="_gapped"
    )
    .filter(
        (pl.col("chromStart") < pl.col("chromEnd_gapped")) &
        (pl.col("chromEnd") > pl.col("chromStart_gapped"))
    )
    .select([
        "chrom",
        "chromStart",
        "chromEnd",
        "name",        # original histone peak name
        "signalValue",
        "pValue",
        "qValue"
    ])
)
```
