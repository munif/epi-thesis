```python
import os
import polars as pl
```

```python
DATASET_PATH = '/group/pmc021/amunif/epi-thesis/workflow/11_HepG2_with_DeepChrome_preprocessing/dataset/'
```

```python
# Define the schema
intersect_schema = pl.Schema({
    # The ensembl_top1.csv columns
    'chrom': pl.String,
    'tss_minus_5k': pl.Int64,
    'tss_plus_5k': pl.Int64, 
    'test_id': pl.String,
    'gene_id': pl.String,
    'chrom_name': pl.String,
    'gene_start': pl.Int64,
    'gene_end': pl.Int64,
    'strand': pl.Int64,
    'gene_name': pl.String,
    'gene_stable_id': pl.String,
    'gene_type': pl.String,
    'tss': pl.Int64,
    'locus': pl.String,
    'orig_start': pl.Int64, 
    'orig_end': pl.Int64,

    # gappedPeak columns
    "chrom_p": pl.String,
    "chromStart_p": pl.Int64, 
    "chromEnd_p": pl.Int64, 
    "name_p": pl.String, 
    "score_p": pl.Float64, 
    "strand_p": pl.String,
    "thickStart": pl.Int64,
    "thickEnd": pl.Int64,
    "itemRgb": pl.Int64,
    "blockCount": pl.Int64,
    "blockSizes": pl.String,
    "blockStarts": pl.String,
    "signalValue": pl.Float64,
    "pValue": pl.Float64,
    "qValue": pl.Float64
})
```

```python
# Load the ensembl dataset using declared schema
df = pl.read_csv(
    os.path.join(DATASET_PATH, 'H3K27me3.bed'),
    separator="\t",
    has_header=False,
    schema = intersect_schema
)
```
