```python
import os
import polars as pl
```

```python
DATASET_PATH = '/group/pmc021/amunif/epi-thesis/workflow/11_HepG2_with_DeepChrome_preprocessing/dataset/'
```

# Read the ensembl dataset

```python
# Define the schema
schema = pl.Schema({
    # The ensembl_top1.csv
    'chrom': pl.String,
    'tss -2kb': pl.Int64,
    'tss +2kb': pl.Int64, 
    'test_id': pl.String,
    'gene_id': pl.String,
    'chrom_name': pl.String,
    'gene_start': pl.Int64,
    'gene_end': pl.Int64,
    'strand': pl.Int64,
    'gene_name': pl.String,
    'gene_stable_id': pl.String,
    'gene_type': pl.String,
    'tss': pl.Int64,
    'locus': pl.String,
    'orig_start': pl.Int64, 
    'orig_end': pl.Int64,
})
```

```python
# Load the ensembl dataset using declared schema
df = pl.read_csv(
    os.path.join(DATASET_PATH, 'ensembl_top1.csv'),
    separator="\t",
    has_header=False,
    new_columns=list(schema.keys()),
    schema_overrides=schema
)
```

```python
df
```

```python
# Remove the previous tss -2k and tss +2k
df = df.drop(["tss -2kb", "tss +2kb"])
```

```python
df
```

```python
# Add the column for +/- 5kb from TSS
df = df.with_columns(
    (pl.col('tss') - 5000).alias('tss_minus_5k'),
    (pl.col('tss') + 5000).alias('tss_plus_5k')
)
```

```python
df
```

```python
# Save to parquet
filename = os.path.join(DATASET_PATH, 'ensembl.parquet')
df.write_parquet(filename)
```

```python
# Test reopen the parquet file
test_df = pl.read_parquet(filename)
test_df
```

```python
# Open the HepG2 expression file
hepg2_df = pl.read_csv(
    os.path.join(DATASET_PATH, 'GSM3718064_HepG2_exp.txt'),
    separator="\t"
)
```

```python
hepg2_df
```

```python
# Join the ensembl and the HepG2 expression
ensembl_HepG2exp_df = df.join(
    hepg2_df,
    left_on = 'gene_id',
    right_on = 'gene_id',
    how = 'left'
)
```

```python
ensembl_HepG2exp_df
```

```python
# Drop the unnecessary column
ensembl_HepG2exp_df = ensembl_HepG2exp_df.drop(['test_id_right', 'locus_right', 'gene', 'sample_1', 'sample_2'])
```

```python
ensembl_HepG2exp_df
```

```python
ensembl_HepG2exp_df.select(['chrom', 'tss_minus_5k', 'test_id', 'value_1'])
```

```python
ensembl_HepG2exp_df.write_parquet(os.path.join(DATASET_PATH, 'ensembl_HepG2exp.parquet'))
```

```python
# Find the column list
ensembl_HepG2exp_df.columns
```

```python
reordered_columns = ['chrom',
 'tss_minus_5k',
 'tss_plus_5k',
 'test_id',
 'gene_id',
 'chrom_name',
 'gene_start',
 'gene_end',
 'strand',
 'gene_name',
 'gene_stable_id',
 'gene_type',
 'tss',
 'locus',
 'orig_start',
 'orig_end',
 'status',
 'value_1',
 'value_2',
 'log2(fold_change)',
 'test_stat',
 'p_value',
 'q_value',
 'significant'
]
```

```python
ensembl_HepG2exp_reordered_df = ensembl_HepG2exp_df.select(reordered_columns)
```

```python
ensembl_HepG2exp_reordered_df
```

```python
# Save the dataset
ensembl_HepG2exp_reordered_df.write_parquet(os.path.join(DATASET_PATH, 'ensembl_HepG2exp.parquet'))
```

```python
# Save the dataset for bedtools intersection
ensembl_HepG2exp_reordered_df.write_csv(os.path.join(DATASET_PATH, 'ensembl_top1.csv'), include_header=False, separator='\t')
```
