```python
def read_gappedPeak(filename):
    # Define the columns
    cols = [
        "chrom", "chromStart", "chromEnd", "name", "score", "strand",
        "thickStart", "thickEnd", "itemRgb", "blockCount", "blockSizes",
        "blockStarts", "signalValue", "pValue", "qValue"
    ]
    
    # Read the file
    gappedpeak_df = pl.read_csv(
        filename,
        separator="\t",
        has_header=False,
        new_columns=cols
    )
    
    # Return the dataframe
    return gappedpeak_df
```

```python
def read_histone(filename):
    # Define the columns
    cols = ["chrom", "chromStart", "chromEnd", "name"]
    
    # Read the histone file
    histone_df = pl.read_csv(
        filename,
        separator="\t",
        has_header=False,
        new_columns=cols
    )

    # Return the dataframe
    return histone_df
```

```python
chroms = histone_df["chrom"].unique().to_list()
overlap_list = []
```

```python
print(chroms)
```

```python
for chrom in chroms:
    print(f"Processing {chrom}...")
    gapped_chr = gapped_df.filter(pl.col("chrom") == chrom)
    histone_chr = histone_df.filter(pl.col("chrom") == chrom)
    
    if gapped_chr.height == 0:
        # No gappedPeak for this chromosome → mark all histone peaks as null
        overlap_list.append(
            histone_chr.with_columns([
                pl.lit(None).alias("signalValue"),
                pl.lit(None).alias("pValue"),
                pl.lit(None).alias("qValue"),
                pl.lit("no_overlap").alias("status")
            ])
        )
        continue
    
    # Cross join per chromosome
    joined = histone_chr.join(gapped_chr, how="cross", suffix="_gapped")
    
    # Vectorized interval overlap
    overlap = joined.filter(
        (pl.col("chromStart") < pl.col("chromEnd_gapped")) &
        (pl.col("chromEnd") > pl.col("chromStart_gapped"))
    ).select([
        "chrom",
        "chromStart",
        "chromEnd",
        "name",           # histone name
        "signalValue",
        "pValue",
        "qValue"
    ])
    
    overlap_list.append(overlap)

# Final result
overlap_df = pl.concat(overlap_list)
```
