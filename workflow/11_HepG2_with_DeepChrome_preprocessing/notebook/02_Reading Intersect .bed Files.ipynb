```python
def read_intersect_bed(filename):
    # Define the columns from bed file
    cols = ["chrom_h", "chromStart_h", "chromEnd_h", "name_h",
            "chrom_p", "chromStart_p", "chromEnd_p", "name_p", "score", "strand",
            "thickStart", "thickEnd", "itemRgb", "blockCount", "blockSizes",
            "blockStarts", "signalValue", "pValue", "qValue"
    ]

    # Read the intersect file
    df = pl.read_csv(
        filename,
        separator="\t",
        has_header=False,
        new_columns=cols
    )

    # Return the dataframe
    return df
```

```python
histone_df = read_intersect_bed(os.path.join(DATASET_PATH, 'H3K9ac.bed'))
```
