```python
import polars as pl
import os
```

```python
DATASET_PATH = '/group/pmc021/amunif/epi-thesis/workflow/11_HepG2_with_DeepChrome_preprocessing/dataset/'
OUTPUT_PATH = '/group/pmc021/amunif/epi-thesis/workflow/11_HepG2_with_DeepChrome_preprocessing/output/'
```

```python
pl_df = pl.read_csv(os.path.join(OUTPUT_PATH, 'test', "*.csv"))
```

```python
pl_df
```

```python
# Add the number of histones column
pl_df = pl_df.with_columns(
     (pl.col("histone_marker").str.count_matches("-") + 1).alias("number_of_histones")
)
```

```python
pl_df
```

```python
# Sort by histone_number (desc) and histone_marker (asc)
pl_df = pl_df.sort(
    by=["number_of_histones", "histone_marker"],
    descending=[True, False]
)
```

```python
pl_df
```

```python
# Add the rank column
pl_df = pl_df.with_columns(
    pl.col('test_accuracy').rank(method='ordinal', descending=True).alias('order_by_test_acc')
)

pl_df
```

```python
# Save to CSV
pl_df.write_csv(os.path.join("HepG2 with DeepChrome preprocessing.csv"))
```
