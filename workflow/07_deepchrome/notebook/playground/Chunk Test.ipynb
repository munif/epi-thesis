```python
import polars as pl
```

```python
# Create a sample DataFrame with 1000 rows
df = pl.DataFrame({
    "id": range(1000),
    "value": [i * 2 for i in range(1000)]
})
```

```python
df
```

```python
# Iterate through the DataFrame in chunks of 100 rows
for i, chunk in enumerate(df.iter_slices(n_rows=100)):
    print(f"Chunk {i+1}:")
    print(chunk.head(5))  # Print first 5 rows of each chunk
    print(f"Chunk size: {len(chunk)}")
    print("---")
    if i == 2:
        break
```
