```python
import pandas as pd
import numpy as np
import time

# Create sample data
df = pd.DataFrame({
    'A': np.random.rand(1000000)
})
array = np.random.rand(1000000)

# Define a function to be applied
def square(x):
    return x ** 2

# Measure time for .apply
start_time = time.time()
df['A_squared'] = df['A'].apply(square)
apply_time = time.time() - start_time

# Measure time for iterating through NumPy array
start_time = time.time()
squared_array_iteration = np.array([x ** 2 for x in array])
iteration_time = time.time() - start_time

# Measure time for NumPy vectorized operation
start_time = time.time()
squared_array_vectorized = array ** 2
vectorized_time = time.time() - start_time

# Print results
print(f"Time taken using .apply: {apply_time} seconds")
print(f"Time taken for iterating through NumPy array: {iteration_time} seconds")
print(f"Time taken for NumPy vectorized operation: {vectorized_time} seconds")
```
