```python
import matplotlib.pyplot as plt
import pandas as pd

from matplotlib.ticker import MultipleLocator
```

```python
# Sample data
categories = ['A', 'B', 'C', 'D']
min_values = [1, 4, 6, 7]
max_values = [5, 8, 9, 10]

# Calculate the widths of the bars
widths = [max_val - min_val for min_val, max_val in zip(min_values, max_values)]

# Create the plot
fig, ax = plt.subplots()

# Add horizontal bars to the plot
ax.barh(categories, widths, left=min_values, color='skyblue')

# Add labels and title
ax.set_ylabel('Category')
ax.set_xlabel('Value Range')
ax.set_title('Horizontal Range Bar Chart Example')

# Show the plot
plt.show()
```

```python
data_df = pd.read_csv("D:\Repo\epi-thesis\dataset\high_count_test\h3k4me3_202405281528.csv", sep="\t", names=['chr', 'start', 'end', 'name'])
data_df.head()
```

```python
# Sample data
categories = data_df['name']
min_values = data_df['start']
max_values = data_df['end']

# Calculate the widths of the bars
widths = [max_val - min_val for min_val, max_val in zip(min_values, max_values)]

# Create the plot
fig, ax = plt.subplots(figsize=(12, 6))

# Set the X range
ax.set_xlim(17719238, 17723238)
ax.xaxis.set_major_locator(MultipleLocator(200))

# Add horizontal bars to the plot
ax.barh(categories, widths, left=min_values, color='skyblue')

# Add gridlines
ax.grid(True, which='both', axis='x', linestyle='--', linewidth=0.5)

# Add labels and title
ax.set_ylabel('Name')
ax.set_xlabel('Bucket')
ax.set_title('H3K4me3 samples')

plt.xticks(rotation=45)

# Show the plot
plt.show()
```
