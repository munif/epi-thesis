```python
import numpy as np
from sklearn.random_projection import GaussianRandomProjection
```

```python
# Generate random data: 100 rows of 4000 elements each
data = np.random.randn(5, 100, 4000)
```

```python
data.shape
```

```python
# Initialize random projector
transformer = GaussianRandomProjection(n_components=200, random_state=42)
```

```python
reduced_data = np.zeros((5, 100, 200))
```

```python
# Apply dimensionality reduction
for i in range(5):
    reduced_data[i] = transformer.fit_transform(data[i])
```

```python
# Verify the shapes
print(f"Original data shape: {data.shape}")
print(f"Reduced data shape: {reduced_data.shape}")
```

```python
# Show samples from each column
print("\nSamples from first row of each column:")
for i in range(5):
    print(f"\nColumn {i}:")
    print(f"Original (first 5 elements): {data[i, 0, :5]}")
    print(f"Reduced (first 5 elements): {reduced_data[i, 0, :5]}")
```
