```python
from sklearn.datasets import make_regression
from sklearn.datasets import make_classification
```

```python
# X, y = make_regression(n_samples=5, n_features=5, noise=1, random_state=42)
X, y = make_classification(n_samples=100, n_features=5, random_state=42)
```

```python
y
```

```python
import numpy as np
import xgboost as xgb
```

```python
seed = 1994
rng = np.random.default_rng(seed)
n_query_groups = 2
qid = rng.integers(0, n_query_groups, size=X.shape[0])
```

```python
qid
```

```python
# Sort the inputs based on query index
sorted_idx = np.argsort(qid)
X = X[sorted_idx, :]
y = y[sorted_idx]
qid = qid[sorted_idx]
```

```python
X
```

```python
y
```

```python
# Create and train the XGBRanker model
ranker = xgb.XGBRanker(
    tree_method="hist",
    lambdarank_num_pair_per_sample=2,
    objective="rank:ndcg",
    lambdarank_pair_method="topk"
)
ranker.fit(X, y, qid=qid)
```

```python
# Make predictions
scores = ranker.predict(X)
```

```python
scores
```

```python
# Sort the relevance scores from most relevant to least relevant
sorted_idx = np.argsort(scores)[::-1]
scores = scores[sorted_idx]
```
