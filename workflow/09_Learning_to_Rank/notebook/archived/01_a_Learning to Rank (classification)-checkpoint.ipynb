```python
from sklearn.datasets import make_classification
import pandas as pd
import numpy as np
import xgboost as xgb
```

```python
# Create a synthetic classification dataset
X, y = make_classification(n_samples=1000, n_features=10, n_informative=5, n_classes=2, random_state=42)
```

```python
# Create a DataFrame with features and add an ID column
df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(X.shape[1])])
df['id'] = np.arange(1000)  # Add ID column
df['label'] = y
```

```python
df
```

```python
# Create a query ID column (for this example, we'll group every 10 samples)
df['qid'] = 0
```

```python
df
```

```python
# Split the data into training and testing sets
train_df = df.sample(frac=0.8, random_state=42)
test_df = df.drop(train_df.index)
```

```python
train_df = train_df.sort_values('qid')
test_df = test_df.sort_values('qid')
```

```python
print(train_df.shape)
print(test_df.shape)
```

```python
# Prepare the data for XGBRanker
X_train = train_df.drop(['id', 'label', 'qid'], axis=1)
y_train = train_df['label']
qid_train = train_df['qid']

# Initialize and train the XGBRanker model
model = xgb.XGBRanker(
    tree_method="hist",
    objective="rank:pairwise",
    eval_metric = 'ndcg',
    random_state=42
)

model.fit(X_train, y_train, qid=qid_train)
```

```python
# Prepare the test data
X_test = test_df.drop(['id', 'label', 'qid'], axis=1)
qid_test = test_df['qid']
```

```python
# Make predictions
test_predictions = []

for qid in qid_test.unique():
    # Get data for the current query
    query_mask = qid_test == qid
    X_query = X_test[query_mask]
    ids_query = test_df['id'][query_mask]
    
    # Predict scores for the current query
    scores = model.predict(X_query)
    
    # Store predictions with their corresponding IDs
    for id, score in zip(ids_query, scores):
        test_predictions.append({'id': id, 'qid': qid, 'predicted_score': score})
```

```python
# Convert predictions to a DataFrame
predictions_df = pd.DataFrame(test_predictions)
```

```python
# Sort predictions by query ID and predicted score (descending)
predictions_df = predictions_df.sort_values(['qid', 'predicted_score'], ascending=[True, False])
predictions_df.sort_values('predicted_score', ascending=False)
```
