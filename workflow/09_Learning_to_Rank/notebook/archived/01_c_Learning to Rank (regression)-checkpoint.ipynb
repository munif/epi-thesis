```python
import numpy as np
import xgboost as xgb
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Generate synthetic ranking data with IDs
def generate_ranking_data(n_queries=50, n_items_per_query=10, n_features=5):
    """
    Create synthetic ranking dataset with unique IDs
    """
    X = []
    y = []
    groups = []
    ids = []
    
    id_counter = 0
    for query in range(n_queries):
        # Generate features for items in this query
        query_X = np.random.rand(n_items_per_query, n_features)
        
        # Create relevance scores 
        query_y = np.random.rand(n_items_per_query)
        query_y.sort()  # Sort to create ranking structure
        
        # Generate unique IDs for each item
        query_ids = np.arange(id_counter, id_counter + n_items_per_query)
        id_counter += n_items_per_query
        
        X.append(query_X)
        y.append(query_y)
        groups.append(np.full(n_items_per_query, query))
        ids.append(query_ids)
    
    # Flatten the lists
    return (
        np.vstack(X),  # Features
        np.concatenate(y),  # Labels
        np.concatenate(groups),  # Groups
        np.concatenate(ids)  # Unique IDs
    )
```

```python
# Generate data
X, y, groups, item_ids = generate_ranking_data()
```

```python
# Split the data
X_train, X_test, y_train, y_test, groups_train, groups_test, ids_train, ids_test = train_test_split(
    X, y, groups, item_ids, test_size=0.2, random_state=42
)
```

```python
# Verify sample sizes
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("groups_train shape:", groups_train.shape)
```

```python
# Create Ranker
ranker = xgb.XGBRanker(
    objective='rank:pairwise',
    learning_rate=0.1,
    n_estimators=100,
    gamma=1.0,
    max_depth=3
)
```

```python
# Get group sizes for training data
train_group_sizes = np.bincount(groups_train)
train_group_sizes
```

```python
# Fit the ranker
ranker.fit(
    X_train, 
    y_train, 
    group=train_group_sizes,
    verbose=True
)
```

```python
# Predict rankings
y_pred = ranker.predict(X_train)
```

```python
# Demonstrate grouping and ranking
print("\nPrediction Details:")
print("Total predictions:", len(y_pred))
print("Prediction scores (top 10):", y_pred[:10])
```

```python
# Create a results DataFrame to track predictions with original IDs
results_df = pd.DataFrame({
    'item_id': ids_train,
    'group': groups_train,
    'features': list(X_train),
    'true_label': y_train,
    'predicted_score': y_pred
})
```
