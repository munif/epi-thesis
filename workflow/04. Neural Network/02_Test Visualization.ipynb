```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
input_file = "/group/pmc021/amunif/epi-thesis/scripts/shell/slurm/workflow/09_Neural_Network/predictions_20240701165538.csv"
```

```python
y = pd.read_csv(input_file)
```

```python
print(y.head())
print(y.shape)
```

```python
# Sort the data
y.sort_values(by=['True Values', 'Predictions'], inplace=True)
print(y.head(10))
```

```python
# Plotting
plt.figure(figsize=(10, 6))
plt.scatter(y.index, y['True Values'], label='y_test', marker='o')
plt.scatter(y.index, y['Predictions'], label='y_pred', marker='x')

plt.xlabel('Row Number')
plt.ylabel('value_1')
plt.yscale('log', base=10)
plt.title('y_test vs y_pred')
plt.legend()
plt.grid(True)

# Save the plot to a file
plt.savefig('nn_ytest_vs_ypred.png')

plt.show()
```
