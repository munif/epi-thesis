```python
import polars as pl
import os
```

```python
DATASET_PATH = '/group/pmc021/amunif/epi-thesis/workflow/15_HepG2 from ENCODE/dataset/'
```

```python
# Open the features file
test_df = pl.read_parquet(os.path.join(DATASET_PATH, 'HepG2_exp_histones.parquet'))
test_df
```

```python
test_df.columns
```

```python
# Select only the gene_id with histones count
gene_histone_count_df = test_df.select([
     'gene_id',
     'H3K9ac_wc',
     'H3K9me3_wc',
     'H3K4me3_wc',
     'H3K27ac_wc',
     'H3K27me3_wc',
     'value_1',
     'value_2'
])
```

```python
gene_histone_count_df
```

```python
gene_histone_count_df = gene_histone_count_df.sort(['value_1'], descending=True)
```
