```python
import pandas as pd
import pyranges as pr
from tqdm.auto import tqdm
import glob
```

```python
import epi_utils as eu
```

```python
chrom_list = pd.read_csv("dataset/chrom_list.csv")["Chromosome"].to_list()
```

# Histone

```python
def read_bed_file(file_name, type):
    df = pr.read_bed(file_name, as_df=True)

    print(f"Processing {type}")

    for chrom in tqdm(chrom_list):
        chrom_df = df[df['Chromosome'] == chrom]
        print(f"Processing {chrom} with {len(chrom_df.index)} data")
        chrom_df.to_csv(f"dataset/histone/{type}/" + chrom + ".bed", sep="\t", index=False, header=False)
    
```

```python
read_bed_file("dataset/HepG2_Male.histone.H3K4me3.peak.bed", "H3K4me3")
read_bed_file("dataset/HepG2_Male.histone.H3K9ac.peak.bed", "H3K9ac")
read_bed_file("dataset/HepG2_Male.histone.H3K27ac.peak.bed", "H3K27ac")
read_bed_file("dataset/HepG2_Male.histone.H3K27me3.peak.bed", "H3K27me3")
read_bed_file("dataset/HepG2_Male.histone.H3K9me3.peak.bed", "H3K9me3")
```

# NCBI Refseq

```python
hg19ncbiRefSeq_df = eu.load_ncbiRefSeq()
display(hg19ncbiRefSeq_df.head())
```

```python
hg19ncbiRefSeq_selected_df = hg19ncbiRefSeq_df[["Chromosome", "Start", "End", "gene_id", "Strand"]]
hg19ncbiRefSeq_selected_df.loc[:, "score"] = 0
hg19ncbiRefSeq_selected_df = hg19ncbiRefSeq_selected_df[['Chromosome', 'Start', 'End', 'gene_id', 'score', 'Strand']]
display(hg19ncbiRefSeq_selected_df.head())
```

```python
cols = list(hg19ncbiRefSeq_selected_df.columns.values)
display(cols)
```

```python
ncbi_chrom_list = hg19ncbiRefSeq_df["Chromosome"].unique().tolist()
display(ncbi_chrom_list)
```

```python
print(f"Splitting into {len(ncbi_chrom_list)} files.")

for chrom in tqdm(ncbi_chrom_list):
    df = hg19ncbiRefSeq_selected_df[hg19ncbiRefSeq_selected_df['Chromosome'] == chrom]
    print(f"Processing {chrom} with {len(df.index)} data")
    df.to_csv(f"dataset/ncbiRefSeq/raw/{chrom}.csv", index=False)
```

```python
# Flatten the glob result
def flatten_concatenation(matrix):
    flat_list = []
    for row in matrix:
        flat_list += row
    return flat_list
```

```python
for chrom in tqdm(chrom_list):
    print(f"Processing chromosome: {chrom}")
    file_list = []
    file_list.append(glob.glob(f"dataset/ncbiRefSeq/raw/{chrom}.csv"))
    file_list.append(glob.glob(f"dataset/ncbiRefSeq/raw/{chrom}_*.csv"))
    flat_file_list = flatten_concatenation(file_list)

    data = []
    for filename in flat_file_list:
        print(f"Processing file: {filename}")
        df = pd.read_csv(filename)
        data.append(df)

    df_out = pd.concat(data, axis=0, ignore_index=False)
    df_out.to_csv(f"dataset/ncbiRefSeq/merged/{chrom}.bed", sep="\t", index=False, header=False)

print("Splitting ncbiRefSeq FINISHED!")
```

# HepG2_exp

```python
# Loading HepG2 data
hepg2_exp = pd.read_csv("dataset/GSM3718064_HepG2_exp.txt", sep="\t")
```

```python
# Splitting chromosome name and the position
hepg2_exp[["chrom", "chromPos"]] = hepg2_exp["locus"].str.split(':', expand=True)
display(hepg2_exp.head())
```

```python
# Splitting start and end position
hepg2_exp[["chromStart", "chromEnd"]] = hepg2_exp["chromPos"].str.split('-', expand=True)
display(hepg2_exp.head())
```

```python
# Correcting the Data Types
hepg2_exp = hepg2_exp.astype({'chromStart': 'int32', 'chromEnd': 'int32'})
```

```python
display(hepg2_exp.head())
```

```python
col_list = (hepg2_exp.columns.values.tolist())
display(col_list)
```

```python
hepg2_exp_selected = hepg2_exp[[ 'chrom',
                        'chromStart',
                        'chromEnd',
                        'test_id',
                        'gene_id',
                        'gene',
                        'locus',
                        'sample_1',
                        'sample_2',
                        'status',
                        'value_1',
                        'value_2',
                        'log2(fold_change)',
                        'test_stat',
                        'p_value',
                        'q_value',
                        'significant',
                        'chromPos']]
```

```python
display(hepg2_exp_selected)
```

```python
hepg2_exp_chrom = hepg2_exp_selected["chrom"].unique().tolist()
```

```python
print(f"Splitting into {len(hepg2_exp_chrom)} files.")

for chrom in tqdm(hepg2_exp_chrom):
    df = hepg2_exp_selected[hepg2_exp_selected['chrom'] == chrom]
    print(f"Processing {chrom} with {len(df.index)} data")
    df.to_csv(f"dataset/HepG2/raw/{chrom}.csv", index=False)
```
